```python
## CONVERTING DATE  COLUMNS ##
```

```python
df['Order Date'] = pd.to_datetime(df["Order Date"])

df['Ship Date'] = pd.to_datetime(df["Ship Date"])
```

```python
df.info()
```

```python
df.head()
```

```python
df["Order Month"] = df["Order Date"].dt.month
df["Order Year"] = df["Order Date"].dt.year
df["Order Day of Week"] = df["Order Date"].dt.dayofweek
```

```python
df
```

```python
## Q1. You need to calculate the monthly sales of the store and identify which month had the highest sales and which month had the lowest sales. ##
```

```python
sales_by_month = df.groupby("Order Month")["Sales"].sum().reset_index()

fig = px.line(
    sales_by_month,
    x="Order Month",
    y="Sales",
    title="Monthly Sales",
    markers=True
)

# Red line
fig.update_traces(
    line=dict(color="red", width=3),
    marker=dict(
        color="white",
        size=10,
        line=dict(color="red", width=3)
    )
)

# Circle + Blue Data Value
fig.add_scatter(
    x=sales_by_month["Order Month"],
    y=sales_by_month["Sales"],
    mode="markers+text",
    text=sales_by_month["Sales"].map(lambda x: f"{x:,.0f}"),
    textposition="middle center",
    marker=dict(
        size=38,
        color="white",
        line=dict(color="red", width=2)
    ),
    textfont=dict(
        color="blue",
        size=11
    ),
    showlegend=False
)

# Background
fig.update_layout(
    plot_bgcolor="#f5f7fa",
    paper_bgcolor="#eaf0f6",
    title=dict(
        text="Monthly Sales",
        x=0.5
    )
)

fig.show()
```

```python
## Q2. Sales of Category ##
```

```python
sales_by_Category=df.groupby('Category')['Sales'].sum().reset_index()

sales_by_Category
```

```python
import plotly.express as px

fig = px.pie(
    sales_by_Category,
    values="Sales",
    names="Category",
    hole=0.5,
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_traces(
    textposition="inside",
    textinfo="percent+label"
)

fig.update_layout(
    title=dict(
        text="Sales by Category",
        x=0.5
    )
)

fig.show()
```

```python
sales_by_Sub_Category=df.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_by_Sub_Category
```

```python
import plotly.express as px

fig = px.bar(
    sales_by_Sub_Category,
    x="Sub-Category",
    y="Sales",
    title="Sales Analysis by Sub-Category",
    text="Sales",
    color="Sub-Category",
    color_discrete_sequence=px.colors.qualitative.Set3
)

# Data labels
fig.update_traces(
    textposition="outside",
    texttemplate="%{text:,.0f}"
)

# Background & layout
fig.update_layout(
    title=dict(
        text="Sales Analysis by Sub-Category",
        x=0.5
    ),
    plot_bgcolor="#f4f6f8",
    paper_bgcolor="#e8eef5",
    xaxis=dict(
        showgrid=False
    ),
    yaxis=dict(
        showgrid=True,
        gridcolor="white"
    ),
    showlegend=False
)

fig.show()
```

```python
 ## Q4.You need to analyze the monthly profit from sales and determine which month had the  highest profit. ##
```

```python
Profit_by_Month=df.groupby('Order Month')['Profit'].sum().reset_index()
Profit_by_Month
```

```python
import plotly.express as px

# Monthly Sales Data
sales_by_month = df.groupby("Order Month")["Sales"].sum().reset_index()

# Line Chart
fig = px.line(
    sales_by_month,
    x="Order Month",
    y="Sales",
    title="Monthly Sales",
    markers=True
)

# Line + Marker + Data Label + Area
fig.update_traces(
    mode="lines+markers+text",

    # Red dotted line
    line=dict(
        color="red",
        width=3,
        dash="dot"
    ),

    # Markers
    marker=dict(
        size=10,
        color="red",
        line=dict(
            color="white",
            width=2
        )
    ),

    # Data Labels
    text=sales_by_month["Sales"].map(
        lambda x: f"{x:,.0f}"
    ),
    textposition="top center",

    # White Data Label
    textfont=dict(
        color="white",
        size=12
    ),

    # Highlight Area
    fill="tozeroy",
    fillcolor="rgba(255, 0, 0, 0.20)"
)

# Chart Layout
fig.update_layout(
    title=dict(
        text="Monthly Sales",
        x=0.5,
        font=dict(
            size=24,
            color="white"
        )
    ),

    # Red Background
    plot_bgcolor="#8B0000",
    paper_bgcolor="#5C0000",

    # Axis
    xaxis=dict(
        title="Order Month",
        color="white",
        showgrid=False
    ),

    yaxis=dict(
        title="Sales",
        color="white",
        showgrid=True,
        gridcolor="rgba(255,255,255,0.25)"
    ),

    # Padding
    margin=dict(
        l=60,
        r=40,
        t=80,
        b=60
    )
)

fig.show()
```
